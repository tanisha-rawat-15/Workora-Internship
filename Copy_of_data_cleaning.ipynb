# Data Type Validation

```python
df.dtypes
```

The data types are appropriate for their respective variables. Numerical variables are stored using integer or floating-point types, while categorical variables are stored as object/string types. No data type conversion was required.

```python
categorical_columns = [
    'Gender',
    'SubscriptionType',
    'ContractType',
    'PaymentMethod',
    'Churn'
]

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

# Check for extra spaces

```python
for col in df.select_dtypes(include='object').columns:
    print(f"\n{col}:")
    print(df[col].apply(lambda x: x != x.strip()).sum())
```

# Check numerical values

```python
print("Age:", df['Age'].min(), "to", df['Age'].max())
print("Tenure:", df['TenureMonths'].min(), "to", df['TenureMonths'].max())
print("Monthly Charges:", df['MonthlyCharges'].min(), "to", df['MonthlyCharges'].max())
print("Total Charges:", df['TotalCharges'].min(), "to", df['TotalCharges'].max())
print("Support Tickets:", df['SupportTickets'].min(), "to", df['SupportTickets'].max())
```

# Validate TotalCharges

```python
calculated_total = (
    df['MonthlyCharges'] * df['TenureMonths']
).round(2)

difference = (
    df['TotalCharges'] - calculated_total
).round(2)

difference
```
